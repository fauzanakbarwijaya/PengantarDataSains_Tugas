```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt

#load data
df = pd.read_csv("calon_tni.csv")

# cetak informasi dari data frame
print("========================== Data Calon TNI ==========================")
print("=" * 68)
print(df)
print("=" * 68)

print()
print()

# cetak statistik deskriptif
print("=" * 68)
print("======================= Statistik Deskriptif =======================")
print(df.describe())

print()
print()

# deteksi masalah dalam data / missing value
print("=" * 42)
print("====== Deteksi Masalah Missing Value =====")
print(df.isnull().sum())

print()
print()

# Deteksi Outlier
print("="*30)
print("====== Deteksi Outlier ======")
Q1 = df["gaji"].quantile(0.25) # 25%
Q3 = df["gaji"].quantile(0.75) # 75%
Q = Q3 - Q1

skrip = "outlier_gaji = df[ df[\"gaji\"] > (Q3 + 1.5 * Q) ]"


print("Kode yang akan dijalankan:")
print(skrip)
print()

outlier_gaji = df[ df["gaji"] > (Q3 + 1.5 * Q) ]

print("Hasil data outlier:")
print(outlier_gaji)

print()
print("====== Outlier Gaji ======")
print(f"Total Outlier Gaji : {len(outlier_gaji)} data") # Menambahkan "Total" agar lebih jelas

print()
print()

# membuat blox-plot
print("="*30)
print("====== Membuat Blox Plot Gaji ======")
df['gaji_juta'] = df['gaji'] / 1000000
gaji_juta = df['gaji_juta']

plt.figure(figsize=(8,6))
plt.boxplot(gaji_juta)
plt.title("Box Plot Gaji Calon TNI")
plt.xlabel("Calon TNI")
plt.ylabel("Gaji (Dalam Juta)")
plt.grid(True, alpha=0.3)
plt.show()
```
